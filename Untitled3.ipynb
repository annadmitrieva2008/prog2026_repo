```python
# ===== Домашня робота 2 =====
# Тест з хімії (

print("Хімічний тест\n")

score = 0

# Питання 1
q1 = input("1) Який хімічний символ має Оксиген?\n   a) O   b) Ox   c) Og\nВаш варіант: ").lower()
if q1 == "a" or q1 == "o":
    score += 1
    print("Правильно!\n")
else:
    print("Неправильно. Правильна відповідь: a) O\n")

# Питання 2
q2 = input("2) Скільки протонів має атом Карбону (C)?\n   a) 4   b) 6   c) 12\nВаш варіант: ").lower()
if q2 == "b" or q2 == "6":
    score += 1
    print("Правильно!\n")
else:
    print("Неправильно. Правильна відповідь: b) 6\n")

# Питання 3
q3 = input("3) Яка з цих речовин є кислотою?\n   a) NaCl   b) HCl   c) CO2\nВаш варіант: ").lower()
if q3 == "b" or q3 == "hcl":
    score += 1
    print("Правильно!\n")
else:
    print("Неправильно. Правильна відповідь: b) HCl\n")

# Підсумок
print("Тест завершено!")
print(f"Ваш результат: {score}/3")

if score == 3:
    print("Відмінно! Ви чудово знаєте хімію.")
elif score == 2:
    print("Добре! Є дрібні помилки.")
else:
    print("Потрібно трохи повторити матеріал.")
```

```python
# ===== Домашня робота 4 =====
# Функція, що отримує список рядків та повертає конкатинацію перших символів
# Тестування на акровірші

def first_letters_concat(lines):
    result = ""
    for line in lines:
        if line:
            result += line[0]
    return result


# Приклад акровірша
acrostic = [
    "Лине спів пташиний в небі раннім,",
    "І росою вмивається трава,",
    "Сонце сходить, світ новий зростає,",
    "Тиша ніжно землю огортає."
]


word = first_letters_concat(acrostic)

print("Отриманий результат:", word)
```

```python
# ===== Домашня робота 5 =====
# Анімація зворотного відліку від 100 до 0
# >60 — зелений, >30 — жовтий, інакше — червоний

import time
import os

# ANSI-коди кольорів
GREEN = "\033[92m"
YELLOW = "\033[93m"
RED = "\033[91m"
RESET = "\033[0m"

for i in range(100, 0, -1):
    # вибір кольору
    if i > 60:
        color = GREEN
    elif i > 30:
        color = YELLOW
    else:
        color = RED

    # очищення екрану (для Windows/macOS/Linux)
    os.system("cls" if os.name == "nt" else "clear")

    print(color + str(i) + RESET)

    time.sleep(0.05)   # невелика затримка для анімаційного ефекту

# фінальне 0
os.system("cls" if os.name == "nt" else "clear")
print(RED + "0" + RESET)
```

```python
# ===== Домашня робота 6 =====
!pip install colorama

import colorama
from colorama import Fore, Back, Style
colorama.init()

import requests
import random
```

```python
def print_guess(word: str, expected: str):
    output = ""

    for pos, letter in enumerate(word):
        if letter == expected[pos]:
            output += Back.GREEN
        elif letter in expected:
            output += Back.YELLOW
        else:
            output += Back.LIGHTBLACK_EX

        output += Fore.WHITE + " " + letter + " " + Style.RESET_ALL

    print(output)


def print_guesses(guesses: list[str], expected: str):
    for guess in guesses:
        print_guess(guess, expected)
```

```python
def get_random_word() -> str:
    response = requests.get("https://api.frontendexpert.io/api/fe/wordle-words")
    return random.choice(response.json()).upper()

expected = get_random_word()
expected
```

```python
MAX_TRIES = 6
guesses = []

print("Починаємо Wordle!")

for attempt in range(MAX_TRIES):

    print(f"\nСпроба {attempt + 1}/{MAX_TRIES}")
    print("Поточна дошка:\n")

    if guesses:
        print_guesses(guesses, expected)
    else:
        print("(порожньо)")

    guess = input("\nВведіть слово з 5 букв: ").strip().upper()

    if len(guess) != 5:
        print("Слово має бути 5 букв!")
        continue

    guesses.append(guess)

    print("\nРезультат цієї спроби:")
    print_guess(guess, expected)

    if guess == expected:
        print("\n🎉 Ви вгадали слово!")
        break

print("\n----- Фінальна дошка -----")
print_guesses(guesses, expected)

if guesses[-1] == expected:
    print("\nРЕЗУЛЬТАТ: ВИГРАЛИ!")
else:
    print(f"\nРЕЗУЛЬТАТ: ПРОГРАШ. Правильне слово: {expected}")
```
